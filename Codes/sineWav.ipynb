```python
import numpy as np
import pandas as pd
from scipy.interpolate import interp1d
```

```python
file_path = '/home/sangheon/Desktop/Pansori_2025_ISMIR/PansoriData/F0/F0_Masked_Split_20/김창환_춘향가_이별가_계면조/김창환 춘향가-이별가_GMjo_40-60.csv'

df = pd.read_csv(file_path, header = None)
df.columns = ['time', 'f0']
time = df['time'].values
f0 = df['f0'].values

valid_idx = ~np.isnan(f0)
interp_function = interp1d(time[valid_idx], f0[valid_idx], kind = 'linear', fill_value = 'extrapolate')

sampling_rate = 44100
duration = time[-1]
t = np.linspace(0, duration, int(sampling_rate * duration))

sine_wave = np.sin(2 * np.pi * interp_function(t) * t)
```
