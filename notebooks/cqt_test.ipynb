```python
from nnAudio.features.cqt import CQT
import librosa
import matplotlib.pyplot as plt
import torch
import numpy as np


song_path = f'{DATA_ROOT}/Audio/0a11a2b7-05-오정숙-흥보가_놀보가_흥보집을_찾아감.wav'
y,sr = librosa.load(song_path, sr=22050)

cqt = CQT(sr=22050, hop_length=512, fmin=32.7, fmax=None, n_bins=84, bins_per_octave=12, filter_scale=1, norm=1, window='hann', center=True, pad_mode='reflect', trainable=False, output_format='Magnitude', verbose=True)


y_tensor = torch.tensor(y).unsqueeze(0)  # shape: [1, samples]
cqt_output = cqt(y_tensor)  # shape: [1, n_bins, time_frames]

# Convert to numpy for plotting
cqt_mag = cqt_output.squeeze(0).numpy()  # shape: [n_bins, time_frames]

# Visualize
plt.figure(figsize=(14, 5))
plt.imshow(
    20 * np.log10(cqt_mag + 1e-6),  # convert to dB
    origin='lower',
    aspect='auto',
    cmap='magma'
)
plt.colorbar(label='dB')
plt.xlabel('Time Frame')
plt.ylabel('CQT Bin')
plt.title('CQT Spectrogram')
plt.tight_layout()
plt.show()

```
